### 2. Deleting Duplicate Data

First we are going to start by identifying any duplicate data in out datasets. After we identify if and/or how many duplicates there are, we will develop a system to delete them strategically. 

Let's start by analyzing the Google Play Store dataset (`GPS_data`). 

**Google Play Store: Identifying Duplicates**

We will create a function that loops through the names of the apps and sorts them into lists. Our first list, `unique_apps_GPS`, will gain every app name once. Our second list, `duplicate_apps_GPS`, will gain every other app in the dataset (duplicates).

```python
unique_apps_GPS = []
duplicate_apps_GPS = []

#loop to seperate unique apps from duplicate apps
for app in GPS_data:
    name = app[0]
    if name in unique_apps_GPS:
        duplicate_apps_GPS.append(name)
    else:
        unique_apps_GPS.append(name)

#print the number of duplicate and unique apps + an example
print(f'Number of duplicate apps: {len(duplicate_apps_GPS)}')
print(f'Number of unique apps: {len(unique_apps_GPS)}')
print('\n')
print(f'Example of duplicate apps: {duplicate_apps_GPS[:5]}')
```

By couting the number of elements in our `duplicate_apps_GPS` list, we see there are 1,181 duplicates. 

**Google Play Store: Systematic Deletion**

*Example:*

Our next task is to systematically delete duplicate entries while keeping the most relevant entry. One way to do this is by looking for differences between the duplicates in the dataset. The below function outputs all of the apps with the name 'Instagram' and gives us its corresponding row number. 

```python
#all apps in the dataset with the name 'Instagram'
for index, app in enumerate(GPS_data):
    if app[0] == 'Instagram':
        print(index)
        print(app)
```

From the output code we can see that the only variation between duplicates is column four: the number of public reviews. The entry with the highest number of reviews (index 2604) would be the best to keep since it contains the most recent data on file. 

*Execution:*

In order to find which duplicates to keep, we have to create a dictionary that holds every app's name and their highest review number. Below we have written a function that loops through every entry in `GPS_data` and puts every new app name in the dictionary. When the dictionary runs into an app name it already has, it compares the number of reviews (`n_reviews`) and takes the highest one. To check the accuracy of our function, we will output the length of the dictionary. If it matches the length of `unique_apps_GPS`, we know our function executed correctly!

```python
reviews_max = {}

#add name with highest number of review to dictionary 
for app in GPS_data:
    name = app[0]
    n_reviews = float(app[3])
    if name in reviews_max and reviews_max[name] < n_reviews:
        reviews_max[name] = n_reviews
    elif name not in reviews_max:
        reviews_max[name] = n_reviews

print(f'Number of rows in reviews_max dictionary: {len(reviews_max)}')
```

We now have a dictionary with 9,659 unique apps and the highest corresponding review value. 

Our next step is to create a list with our unique apps and all of their corresponding information so later we can use this cleansed list to effectively analyse the dataset. 

*Execution: Issue*

We will loop through every row in the origonal Google Play Store database, `GPS_data`. As we run through every row we will compare the number of reviews, `n_reviews`, to our dictionary, `reviews_max`. If the number of reviews for the row we are analyzing is equal to the highest number of reviews listed in `reviews_max` the entire row in the Google Play Store database (13 columns of data) will be added to the list `cleaned_GPS_data`.

However, there is a catch. 
What if our `reviews_max` dictionary hold this value {'Instagram': '66577466'} and there are two rows in the Google Play Store database with the name 'Instagram' and the number of reviews '66577466'?

Our current approach would allow both rows to be added to `cleaned_GPS_data` and introduce duplicates back into our database. 

*Execution: Solution*

To solve this problem we will introduce another list: `already_added`. When we loop through Google Play Store row-by-row, if the number of reviews match our dictionary and the name of the app is not in our list `already_added` we will add the row to `cleaned_GPS_data` and the name to our list `already_added`. Now when other app has the same number of reviews as our dictionary but its name is already in our `already_added` list, it won't be added to the `cleaned_GPS_data` list. 

```python
cleaned_GPS_data = []
already_added = []

#pull rows from original dataset where the number of reviews match the values in 'reviews_max' dictionary 
#only if name isn't in 'already_added' list (first time n_reviews match 'reviews_max' name is added to 'already_added')  
for app in GPS_data:
    name = app[0]
    n_reviews = float(app[3])
    if (n_reviews == reviews_max[name]) and (name not in already_added):
        cleaned_GPS_data.append(app[:])
        already_added.append(name)

print(f'Number of rows in cleaned_GPS_data list: {len(cleaned_GPS_data)}')     
```

Now `cleaned_GPS_data` is a list of lists with the expected 9,659 rows and no duplicates!

### 3. Deleting Non-English Apps

Remember our company develops apps for English-speaking audiences so we need to discard any non-English apps for our analysis. Before we jump into writing code, we need to decipher what ways the computer can differentiate English letters (e.g. a, b, etc.) and symbols (e.g. *, /, etc.) from non-English letters and symbols. 

*Computer Characters:* 

Unlike people, the computer "reads" letters or characters in numbers. For example, character "A" is 65 to the computer and '爱' is 29,233. Lucky for us, most of the English letters and symbols are within the range 0 to 127 according to the ASCII (American Standard Code for Information Interchange) system. This means we can loop through the names of our apps and test the letters against a range of 0 to 127 using the `ord()` function. 

*Exploration:*

The `ord()` function is a built-in function that converts any character to its ASCII number. The function we create (`english`) will take a single argument (app_name) and loop through every letter in a given app name. If the letter has an ASCII value that is greater than 127, the function will return False. If the letter is less or equal to 127, the function will return True. Below, we will test this out. 

```python
#function that returns a boolean based on whether the characters have a number value greater than 127
def english(app_name):
    for letter in app_name:
        if ord(letter) > 127:
            return False
    return True

#print examples
print(english('Instagram'))
print(english('爱奇艺PPS -《欢乐颂2》电视剧热播'))
print(english('Docs To Go™ Free Office Suite'))
print(english('Instachat 😜'))
```

As we can see above, our function did not output quite how we expected. Two of the English app names returned False. This is because emojis and characters like ™ correspond to a number value greater than 127. For our purposes, we will allow every word to have 3 characters with a ASCII value of over 127. All other app names will be discarded. 

To do this, we will create a tracker that counts the number of times a word has a character with a value greater than 127. If the word has less than 3 characters with values over 127, the function will return True. If the word has more than 3 characters with values over 127, the function will return False. 

*Execution:*

```python
#return a boolean based on whether a function has three character with number values greater than 127
def english(app_name):
    count = 0
    for letter in app_name:
        if ord(letter) > 127:
            count += 1

    if count > 3:
        return False
    else:
        return True

#print examples
print(english('Instagram'))
print(english('爱奇艺PPS -《欢乐颂2》电视剧热播'))
print(english('Docs To Go™ Free Office Suite'))
print(english('Instachat 😜'))
```

Our function now outputs the anticipated results. While we may still loose some English apps using this function, most of them will remain intact for our analysis. 

**Google Play Store & Apple Store: Non-English App Deletion**

## Most Common Apps By Genre

The goal of the company is to create apps that have high user engagement because the company relies on revenue from in-app adds. So, the more engagement with the app also means more engagement with the adds. 

The company tests their new apps by these three steps:
1) Build a simple Android app and add it to the Google Play Store (by starting with a minial app the company is minimizing their risk and overhead).
2) If the app does well on the Google Play Store, the app is further developed for use on Android devices.
3) If the app is profitable after 6 months, the company builds an IOS version and adds it to the Apple Store. 

Since the final goal is to have one app that is profitable on both Android and IOS devices, we will identify the most popular genre of app on the Google Play Store and Apple Store databases. 

**Google Play Store:**

To start, we will create a dictionary that counts the number of times a distinct category name arises. 

*Loop 1:*
In the function below, the function loops though every row of data in our cleaned Google Play Store dataset `GPS_final_clean`. For every row, the function analyzes the category listed in the category column (category = row[1]). If the category listed is not already in our dictionary, we add the category name and associate it with a value of 1. If, the category listed is already in our dictionary, we simply add 1 to the value already associated with it. Once our function has run through all the rows in the dataset, our dictionary will contain grouped categories and the number of times that category appears in the dataset (`cat_dic_GPS`). 

*Loop 2:*
The next task is to convert all of the values in that dictionary to percentages. Our next function will loop through `cat_dic_GPS`, and divide the value associated with every grouped category by the total number of rows in the dataset and then multiply by 100. Our final dictionary (`percent_dic`) will contain every grouped category with the percentage of frequency in the dataset. 

See below the function and output result. 

```python
from pprint import pprint
cat_dic_GPS = {}
total = 0 

#loop 1: for category count dictionary 
for row in GPS_final_clean:
    total += 1
    category = row[1]
    if category in cat_dic_GPS:
        cat_dic_GPS[category] += 1
    else:
        cat_dic_GPS[category] = 1

#loop 2: for category percent dictionary
percent_dic = {}
for key in cat_dic_GPS:
    percentage = (cat_dic_GPS[key] / total) * 100
    percent_dic[key] = percentage

print(f'Counted number of rows in the cleaned Google Play Store dataset: {total}')
pprint(cat_dic_GPS)
```

*(Note: (a) pprint is an imported module to make the dictionary more readible and (b) we printed the 'cat_dic_GPS' dictionary not the percentages dictionary)*

The code above grouped every distinct category seperately (key). The number of times that category appeared is listed next to it (value). 

While the code above it good but... 
1) it does not allow us to sort based on the number of occurances or
2) allow us to easily use another dataset in its place.

To addess the issues...
1) For the first issue, we will have to convert every key-value pair to a tuple so it can be sorted by the `sorted()` function.
2) For the second issue, we will create a function (`freq_table`) that takes two arguments: (a) one for the dataset of interest and (b) one for the column in the dataset. 

If you will notice, the code within the function `freq_table` is identical to the code written above. The only difference is we renamed most elements to strengthen comprehension and added the two arguments for reusibility.

```python
#function used in previous code cell (altered for reusibility)
def freq_table(dataset, index): # dataset is list of lists # index is integer
    dictionary = {}
    total = 0

#loop 1: for category count dictionary     
    for row in dataset:
        total += 1
        column = row[index]
        if column in dictionary:
            dictionary[column] += 1
        else:
            dictionary[column] = 1 

#loop 2: for category percent dictionary    
    dict_percent = {}
    for key in dictionary:
        percentage = (dictionary[key] / total) * 100
        dict_percent[key] = percentage

    return dict_percent

def display_table(dataset, index):
    table = freq_table(dataset, index)
    table_display = []

#turn key-value-pairs into tuples    
    for key in table:
        key_val_as_tuple = (table[key], key)
        table_display.append(key_val_as_tuple)

#sort the tuples by percentage in descending order    
    table_sorted = sorted(table_display, reverse = True)
    for entry in table_sorted:
        print(entry[1], ':', entry[0])
```

The last function on the page `display_table` is a function that takes the grouped categories and their associated percentages (key, value) and turns all the key-value pairs into tuples, adds those pairs to a list and then sorts the list from highest to lowest percentages. We will use `display_table` to output and analyze genres in each database.

Since the goal of the company is to design apps that have high user engagement, let's analyze the average number of installs per app per genre. The first code block below executes this function and sorts the results for the `GooglePlayStore`.

**Average Installs Per App Genre:**

```python
android_dict = freq_table(GPS_final_clean, 1)

#loop that adds the average installs per genre into the dictionary
dictionary = {}
for genre in android_dict:
    installs_per_genre = 0
    apps_per_genre = 0
    for app in GPS_final_clean:
        app_genre = app[1]
        if app_genre == genre:
            installs = app[5]
            installs = installs.replace(',', '')
            installs = installs.replace('+', '')
            installs_per_genre += float(installs)
            apps_per_genre += 1
            avg_installs = (installs_per_genre / apps_per_genre )
            dictionary[genre] = avg_installs

#convert key-value pairs into tuples
table_display = []
for key in dictionary:
        key_val_as_tuple = (dictionary[key], key)
        table_display.append(key_val_as_tuple)

#sort based on highest average installs
table_sorted = sorted(table_display, reverse = True)
for entry in table_sorted:
    print(f'{entry[1]} : {entry[0]}')
```

**Spread of Installments Per Genre:**

```python
from pprint import pprint
list1 = ['COMMUNICATION', 'VIDEO_PLAYERS', 'SOCIAL']
dic_2 = {}

#loop that adds the number apps with 'x' amount of installs to 'dict_2'
for genre in list1:
    dic_2[genre] = {}
    for app in GPS_final_clean:
        if app[1] == genre:
            installs = app[5]
            installs = float(installs.replace(",", "").replace("+", ""))
            if installs in dic_2[genre]:
                dic_2[genre][installs] += 1
            else:
                dic_2[genre][installs] = 1

#pprint orders dictionary in ascending order based on index[0]
pprint(dic_2)
```
